# Plutus Debugger: Gas Profiling (Plutus V2 & V3)

This tutorial demonstrates how to profile your Plutus contracts to minimize ExUnit consumption.

```haskell
{-# INLINABLE expensiveComputation #-}
expensiveComputation :: Integer -> Integer
expensiveComputation 0 = 1
expensiveComputation n = n * expensiveComputation (n - 1)
```

### Running the Profiler
Open the Command Palette (`Ctrl+Shift+P` or `Cmd+Shift+P`) and type **Plutus Debugger: Profile Gas**.
The extension will execute the script and generate an flamegraph output in the side panel showing the Memory and CPU ExUnits consumed by `expensiveComputation`.
