# Plutus Debugger: Breakpoint Debugging Tutorial

Welcome to the interactive tutorial for breakpoint debugging in Plutus using the VSCode Plugin.
In this notebook, we'll cover how to define a validator, attach the debugger, and step through the execution.

```haskell
-- Plutus Core Boilerplate
{-# INLINABLE mkValidator #-}
mkValidator :: BuiltinData -> BuiltinData -> BuiltinData -> ()
mkValidator datum redeemer context =
    if redeemer == BuiltinData 42
    then () -- Success
    else error () -- Failure
```

### Setting Breakpoints
In your VSCode editor, simply click to the left of the line number on `if redeemer == BuiltinData 42`.
A red dot will appear, indicating a breakpoint.

When you start the debugger via `F5`, execution will pause here, allowing you to inspect the `datum`, `redeemer`, and `context`.
